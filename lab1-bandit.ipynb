```python
%pylab inline
from scipy.stats import norm
```

```python
# Number of bandit arms
K = 10 # @param {"type":"integer"}
# Number of trials (time steps)
T = 1000  # @param {"type":"integer"}

# Random seed
seed = 999 # @param {"type":"integer"}
std_upper = 5 # @param {"type":"integer"}
```

```python
# Stattionary probability distribution for K bandits
np.random.seed(seed)
bandit_mean = 10 * np.random.normal(size=K)
bandit_std = np.random.uniform(1, std_upper, size=K)




###### Plot disttributions #######
fig, axes = plt.subplots(1, K, figsize=(2*K, 4), sharey=True)

x_min = bandit_mean.min() - 4 * bandit_std.max()
x_max = bandit_mean.max() + 4 * bandit_std.max()
x = np.linspace(x_min, x_max, 500)

for i, ax in enumerate(axes):
    y = norm.pdf(x, loc=bandit_mean[i], scale=bandit_std[i])
    ax.plot(y, x, color="C0")  # vertical orientation
    ax.fill_betweenx(x, 0, y, alpha=0.2, color="C0")
    ax.axhline(bandit_mean[i], color="r", linestyle="--", linewidth=1)  # mean
    ax.set_title(f"Arm {i}")
    ax.set_xlabel("Density")
    if i == 0:
        ax.set_ylabel("Reward")
    ax.grid(True, linestyle="--", alpha=0.5)
plt.suptitle("Unknown Reward Distributions for K Bandits", y=1.02)
```

```python
np.random.seed(seed)

# Rewards
rewards = []

# For T trials, chose a random arm
for t in range(T):
  # Action: Choose a random arm
  a = np.random.randint(0, K-1)

  # Reward
  r = np.random.normal(bandit_mean[a], bandit_std[a])
  rewards.append(r)


# Total reward
total = sum(rewards)
avg_reward = np.cumsum(rewards) / np.arange(1, len(rewards) + 1)

# Plot results
plot(rewards, 'b--o', label='Rewards')
plot(avg_reward, '-r', label='Average Reward')
legend()
title("Rewards")
plt.title("Total Reward %.2f"%total)
xlabel("Time step (trial)")
ylabel("Reward")
```

The average of n independent Gaussian random variables is also a Gaussian random variable. The resulting Gaussian will have a mean equal to the average of the individual means (Σμᵢ)/n, and a variance equal to the average of the individual variances (Σσᵢ²)/n.

## Exercise 2: Greedy strategy


1. Implement the greedy strategy: Always choose the arm with the highest estimated average reward.

2. Implement the Epsilon-greedy strategy: with a small probability epsilon, choose a random arm; otherwise, choose the greedy arm.

3. [Optional for 337] Implement Thompson Sampling: Maintain a probability distribution over the expected reward for each arm and sample from these distributions to choose an arm.



```python
# Greedy
np.random.seed(seed)


# Rewards
rewards = []

# Find the best action among the first five trials
action_rewards = [[] for _ in range(K)]

for t in range(0, K):
    a = t
    reward = np.random.normal(bandit_mean[a], bandit_std[a])
    rewards.append(reward)
    action_rewards[a].append(reward)

best_a = np.argmax([np.mean(action_reward) for action_reward in action_rewards])
print(f"Best arm: {best_a}")

# Action: Choose the best action found in the first trials
for t in range(0, T):
    # Action: Choose the best arm
    a = best_a

    # Reward
    r = np.random.normal(bandit_mean[a], bandit_std[a])
    action_rewards[a].append(r)

    best_a = np.argmax([np.mean(action_reward) for action_reward in action_rewards])
    if best_a != a:
        print(f"Best arm: {best_a}")
    rewards.append(r)


# Total reward
total = sum(rewards)

# Plot results
plot(rewards, 'b--o', label='Rewards')
plt.title("Total Reward %.2f"%total)
xlabel("Time step (trial)")
ylabel("Reward")

```

```python
# Epsilon-greedy
np.random.seed(seed)


# Rewards
rewards = []
action_rewards = [[0] for _ in range(K)]

epsilon = 0.1
# Action: Choose the best action found in the first five trials

for t in range(0, K):
    a = t
    reward = np.random.normal(bandit_mean[a], bandit_std[a])
    rewards.append(reward)
    action_rewards[a].append(reward)

for t in range(K, T):
    # Action: Choose a random arm
    if np.random.rand() < epsilon or best_a == -1:
        a = np.random.randint(0, K-1)
        print(f"Random arm: {a}")
    else:
        a = np.argmax([np.mean(action_reward) for action_reward in action_rewards])

    # Reward
    r = np.random.normal(bandit_mean[a], bandit_std[a])
    action_rewards[a].append(r)
    rewards.append(r)


# Total reward
total = sum(rewards)

# Plot results
plot(rewards, 'b--o', label='Rewards')
plt.title("Total Reward %.2f"%total)
xlabel("Time step (trial)")
ylabel("Reward")
```
